The fitted model is:
Model: Y = β0 + β1·GPA + β2·IQ + β3·Level + β4·(GPA×IQ) + β5·(GPA×Level)

Coefficients: β0=50, β1=20, β2=0.07, β3=35, β4=0.01, β5=-10.

(a) Which answer is correct?
For a college graduate, Level=1; for a high-school graduate, Level=0.

The college-minus-high-school difference is: 35 − 10(GPA)
Setting this equal to zero gives GPA = 3.5.
GPA < 3.5: college graduates earn more.
GPA = 3.5: predicted salaries are equal.
GPA > 3.5: high-school graduates earn more.

```python
beta3 = 35
beta5 = -10

for gpa in [3.0, 3.5, 4.0]:
    difference = beta3 + beta5 * gpa
    print(f'GPA = {gpa}: College - High School = {difference:.1f} thousand dollars')

print('\nCrossover point: GPA = 3.5')
print('Answer: (iii)')
```

(b) Predict the salary of a college graduate with IQ = 110 and GPA = 4.0
GPA = 4, IQ = 110, Level = 1.

GPA×IQ = 440 and GPA×Level = 4.

Predicted salary = 50 + 20(4) + 0.07(110) + 35(1) + 0.01(440) − 10(4) = 137.1 thousand dollars.

Answer: $137,100.

```python
beta0, beta1, beta2, beta3, beta4, beta5 = 50, 20, 0.07, 35, 0.01, -10
GPA = 4.0
IQ = 110
Level = 1

GPA_IQ = GPA * IQ
GPA_Level = GPA * Level

salary_thousands = (beta0 + beta1*GPA + beta2*IQ + beta3*Level
                    + beta4*GPA_IQ + beta5*GPA_Level)
salary_dollars = salary_thousands * 1000

print(f'GPA × IQ = {GPA_IQ}')
print(f'GPA × Level = {GPA_Level}')
print(f'Predicted salary = {salary_thousands:.1f} thousand dollars')
print(f'Predicted salary = ${salary_dollars:,.0f}')
```
