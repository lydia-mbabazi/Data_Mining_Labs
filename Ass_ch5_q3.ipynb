Conceptual question: reviewing k-fold cross-validation.

**(a)** Explain how k-fold cross-validation is implemented.

The observations are randomly divided into k groups (folds) of roughly equal size. One fold at a time is treated as the validation set, and the model is fit on the remaining k − 1 folds. The fitted model is then used to predict the held-out fold, and the error (e.g. MSE) is recorded. This is repeated k times, once per fold, giving k error estimates MSE₁, ..., MSEₖ, which are averaged to give the k-fold CV estimate:

CV(k) = (1/k) Σ MSEᵢ

**(b)** Advantages and disadvantages of k-fold CV relative to:

**i. The validation set approach**

Advantages of k-fold CV:
- Every observation is used for both training and validation (across the k rounds), so the estimate is far less variable than a single random split.
- Since each training fold uses roughly (k−1)/k of the data (more than the ≈ half used by a single validation split), k-fold CV has less bias — it tends not to overestimate the test error as much.

Disadvantage: k-fold CV requires fitting the model k times instead of once, so it is more computationally expensive than a single validation set split.

**ii. LOOCV**

Advantages of k-fold CV (with k < n) over LOOCV:
- Computationally cheaper: k fits (typically k = 5 or 10) instead of n fits.
- Lower variance: LOOCV's n fitted models are trained on almost identical, heavily overlapping data, so their outputs are highly correlated, which inflates the variance of the averaged estimate. k-fold CV's folds overlap less, so its estimate has lower variance.

Disadvantage of k-fold CV relative to LOOCV:
- More bias: LOOCV trains on n − 1 observations each time (almost the full data set), so it is nearly unbiased for the test error. k-fold CV trains on fewer observations per fit, so its estimate is somewhat more biased upward.

In practice, k = 5 or k = 10 is preferred as a bias-variance compromise between the validation set approach and LOOCV.
