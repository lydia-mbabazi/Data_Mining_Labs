```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

# ============================================================
# 3. Bias-Variance Decomposition
# ============================================================

# Flexibility of the statistical learning method
# 0 = very inflexible, 10 = very flexible
flexibility = np.linspace(0, 10, 500)

# ------------------------------------------------------------
# 1. Squared Bias
# ------------------------------------------------------------
# Bias decreases as flexibility increases.
squared_bias = 3.5 * np.exp(-0.45 * flexibility) + 0.20

# ------------------------------------------------------------
# 2. Variance
# ------------------------------------------------------------
# Variance increases as flexibility increases.
variance = 0.10 * np.exp(0.48 * flexibility)

# ------------------------------------------------------------
# 3. Training Error
# ------------------------------------------------------------
# Training error decreases as flexibility increases.
training_error = 4.0 * np.exp(-0.45 * flexibility) + 0.10

# ------------------------------------------------------------
# 4. Bayes / Irreducible Error
# ------------------------------------------------------------
# Irreducible error is constant regardless of flexibility.
bayes_error = np.full_like(flexibility, 0.45)

# ------------------------------------------------------------
# 5. Test Error
# ------------------------------------------------------------
# Test error is affected by both bias and variance.
# It typically decreases at first and then increases.
test_error = 0.45 + 0.50 * squared_bias + 0.25 * variance

```

```python

# ============================================================
# PART (a): Plot the five curves
# ============================================================

plt.figure(figsize=(12, 7))

plt.plot(
    flexibility,
    squared_bias,
    label="Squared Bias",
    linewidth=2
)

plt.plot(
    flexibility,
    variance,
    label="Variance",
    linewidth=2
)

plt.plot(
    flexibility,
    training_error,
    label="Training Error",
    linewidth=2
)

plt.plot(
    flexibility,
    test_error,
    label="Test Error",
    linewidth=3
)

plt.plot(
    flexibility,
    bayes_error,
    label="Bayes / Irreducible Error",
    linestyle="--",
    linewidth=2
)

# Labels and title
plt.xlabel(
    "Flexibility of Statistical Learning Method",
    fontsize=12
)

plt.ylabel(
    "Error",
    fontsize=12
)

plt.title(
    "Bias-Variance Decomposition",
    fontsize=15
)

# Add legend
plt.legend(fontsize=10)

# Add grid
plt.grid(alpha=0.3)

# Set limits
plt.xlim(0, 10)
plt.ylim(0, 6)

# Display plot
plt.tight_layout()
plt.show()

```

```python

# ============================================================
# PART (b): Explanation of the five curves
# ============================================================

print("\n" + "=" * 70)
print("PART (b): EXPLANATION OF THE FIVE CURVES")
print("=" * 70)

print("""
1. SQUARED BIAS
----------------
Squared bias decreases as the flexibility of the statistical
learning method increases.

A less flexible model makes strong assumptions about the
relationship between the predictors and the response. Therefore,
it may be too simple to capture the true relationship, resulting
in high bias.

As flexibility increases, the model becomes better able to capture
complex relationships in the data. Therefore, squared bias decreases.

In short:
More flexibility -> Lower bias
""")


print("""
2. VARIANCE
-----------
Variance increases as the flexibility of the model increases.

A less flexible model is relatively stable and does not change
much when the training data changes.

A highly flexible model can respond strongly to small changes in
the training data. Such a model may fit not only the true pattern
but also random noise in the training data.

Therefore, variance increases with flexibility.

In short:
More flexibility -> Higher variance
""")


print("""
3. TRAINING ERROR
-----------------
Training error decreases as flexibility increases.

A more flexible model has greater ability to fit the observations
in the training dataset.

For example, a very flexible model can closely follow individual
training observations and may eventually achieve a very small
training error.

Therefore, training error generally decreases as flexibility
increases.

In short:
More flexibility -> Lower training error
""")


print("""
4. TEST ERROR
-------------
Test error usually has a U-shaped curve.

When the model is not flexible enough, it is too simple and
underfits the data. This produces high bias and therefore high
test error.

As flexibility increases, the model captures more of the true
relationship in the data. Bias decreases and test error also
decreases.

However, if the model becomes excessively flexible, it begins
to fit noise and random characteristics of the training data.
This is called overfitting.

At this point, variance becomes large and test error starts to
increase.

Therefore:
Low flexibility -> High bias -> High test error
Optimal flexibility -> Lowest test error
High flexibility -> High variance -> High test error
""")


print("""
5. BAYES / IRREDUCIBLE ERROR
----------------------------
Bayes error, also called irreducible error, remains approximately
constant regardless of the flexibility of the statistical learning
method.

Some amount of error is caused by random noise or factors that
cannot be observed or measured.

Even a perfect statistical learning method cannot completely
remove this error.

Therefore, increasing model flexibility does not reduce the
Bayes or irreducible error.

In short:
More flexibility -> Bayes error remains constant
""")


print("""
============================================================
BIAS-VARIANCE DECOMPOSITION
============================================================

Expected Test Error =
    Squared Bias
    + Variance
    + Irreducible Error

The main trade-off is between bias and variance.

As flexibility increases:
    - Squared bias decreases
    - Variance increases
    - Training error decreases
    - Test error first decreases and then increases
    - Bayes/irreducible error remains constant

The goal is to choose a level of flexibility that gives a
low test error without causing excessive overfitting.
""")
```
